```python
import zipfile

zip_path = "/content/drive/MyDrive/Link file one/PC/Games/Waterpark-Simulator-AnkerGames.zip"

with zipfile.ZipFile(zip_path, "r") as z:
    url_files = [name for name in z.namelist() if name.lower().endswith(".url")]

print("Found", len(url_files), ".URL file(s):")
for name in url_files:
    print(name)
```
